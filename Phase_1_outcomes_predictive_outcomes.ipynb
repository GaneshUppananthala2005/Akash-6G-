```python
# ============================================================
# PROJECT AKASH-6G
# IEEE PAPER FIGURE GENERATION
# Physics-Informed LEO Satellite Beam Selection
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# ------------------------------------------------------------
# 1. FILE
# ------------------------------------------------------------

EXCEL_FILE = "AKASH_6G_Accurate_Physics_Informed_Beam_Dataset_50000.xlsx"

OUTPUT_DIR = "AKASH6G_IEEE_Figures"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ------------------------------------------------------------
# 2. LOAD DATA
# ------------------------------------------------------------

# Use all_data if available.
# This avoids plotting the same physical trajectory four times.

df = pd.read_excel(
    EXCEL_FILE,
    sheet_name="all_data"
)

print("Dataset loaded successfully")
print("Rows:", len(df))
print("Columns:", len(df.columns))

print(df.columns.tolist())

# ------------------------------------------------------------
# 3. IEEE FIGURE SETTINGS
# ------------------------------------------------------------

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "DejaVu Serif"],
    "font.size": 9,
    "axes.labelsize": 9,
    "axes.titlesize": 9,
    "legend.fontsize": 8,
    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
    "figure.dpi": 150
})

# ------------------------------------------------------------
# Helper function
# ------------------------------------------------------------

def save_ieee(fig, filename):

    path_png = os.path.join(
        OUTPUT_DIR,
        filename + ".png"
    )

    path_pdf = os.path.join(
        OUTPUT_DIR,
        filename + ".pdf"
    )

    fig.savefig(
        path_png,
        dpi=600,
        bbox_inches="tight"
    )

    fig.savefig(
        path_pdf,
        bbox_inches="tight"
    )

    print("Saved:", path_png)
    print("Saved:", path_pdf)


# ============================================================
# FIGURE 1
# DOPPLER SHIFT VS TIME
# ============================================================

fig, ax = plt.subplots(
    figsize=(3.45, 2.5)
)

ax.plot(
    df["Time_ms"],
    df["Doppler_kHz"],
    linewidth=1.0
)

ax.axhline(
    0,
    linestyle="--",
    linewidth=0.8
)

ax.set_xlabel("Time (ms)")
ax.set_ylabel("Doppler Shift (kHz)")

ax.grid(
    True,
    linestyle=":",
    linewidth=0.5,
    alpha=0.7
)

fig.tight_layout()

save_ieee(
    fig,
    "Fig1_Doppler_Shift"
)

plt.show()


# ============================================================
# FIGURE 2
# SLANT RANGE VS ELEVATION
# ============================================================

fig, ax1 = plt.subplots(
    figsize=(3.45, 2.5)
)

ax1.plot(
    df["Time_ms"],
    df["Slant_Range_km"],
    linewidth=1.0,
    label="Slant Range"
)

ax1.set_xlabel("Time (ms)")
ax1.set_ylabel("Slant Range (km)")

ax1.grid(
    True,
    linestyle=":",
    linewidth=0.5,
    alpha=0.7
)

ax2 = ax1.twinx()

ax2.plot(
    df["Time_ms"],
    df["Elevation_Angle_deg"],
    linestyle="--",
    linewidth=1.0,
    label="Elevation"
)

ax2.set_ylabel("Elevation Angle (deg)")

fig.tight_layout()

save_ieee(
    fig,
    "Fig2_SlantRange_Elevation"
)

plt.show()


# ============================================================
# FIGURE 3
# FOUR-BEAM RSRP COMPARISON
# ============================================================

fig, ax = plt.subplots(
    figsize=(3.45, 2.5)
)

for beam in range(4):

    column = f"beam{beam}_RSRP_dBm"

    if column in df.columns:

        ax.plot(
            df["Time_ms"],
            df[column],
            linewidth=0.9,
            label=f"Beam {beam}"
        )

ax.set_xlabel("Time (ms)")
ax.set_ylabel("RSRP (dBm)")

ax.grid(
    True,
    linestyle=":",
    linewidth=0.5,
    alpha=0.7
)

ax.legend(
    frameon=False,
    ncol=2
)

fig.tight_layout()

save_ieee(
    fig,
    "Fig3_FourBeam_RSRP"
)

plt.show()


# ============================================================
# FIGURE 4
# FOUR-BEAM SINR COMPARISON
# ============================================================

fig, ax = plt.subplots(
    figsize=(3.45, 2.5)
)

for beam in range(4):

    column = f"beam{beam}_SINR_dB"

    if column in df.columns:

        ax.plot(
            df["Time_ms"],
            df[column],
            linewidth=0.9,
            label=f"Beam {beam}"
        )

ax.set_xlabel("Time (ms)")
ax.set_ylabel("SINR (dB)")

ax.grid(
    True,
    linestyle=":",
    linewidth=0.5,
    alpha=0.7
)

ax.legend(
    frameon=False,
    ncol=2
)

fig.tight_layout()

save_ieee(
    fig,
    "Fig4_FourBeam_SINR"
)

plt.show()


# ============================================================
# FIGURE 5
# BEAM SELECTION / HANDOVER
# ============================================================

fig, ax = plt.subplots(
    figsize=(3.45, 2.2)
)

ax.step(
    df["Time_ms"],
    df["Current_Beam"],
    where="post",
    linewidth=1.0
)

ax.set_xlabel("Time (ms)")
ax.set_ylabel("Selected Beam")

ax.set_yticks([0, 1, 2, 3])

ax.set_ylim(
    -0.5,
    3.5
)

ax.grid(
    True,
    linestyle=":",
    linewidth=0.5,
    alpha=0.7
)

fig.tight_layout()

save_ieee(
    fig,
    "Fig5_Beam_Handover"
)

plt.show()


# ============================================================
# FIGURE 6
# OPTIMAL BEAM LABEL
# ============================================================

fig, ax = plt.subplots(
    figsize=(3.45, 2.2)
)

ax.step(
    df["Time_ms"],
    df["Optimal_Beam_Label"],
    where="post",
    linewidth=1.0
)

ax.set_xlabel("Time (ms)")
ax.set_ylabel("Optimal Beam")

ax.set_yticks([0, 1, 2, 3])

ax.set_ylim(
    -0.5,
    3.5
)

ax.grid(
    True,
    linestyle=":",
    linewidth=0.5,
    alpha=0.7
)

fig.tight_layout()

save_ieee(
    fig,
    "Fig6_Optimal_Beam"
)

plt.show()


# ============================================================
# FIGURE 7
# THROUGHPUT
# ============================================================

fig, ax = plt.subplots(
    figsize=(3.45, 2.5)
)

ax.plot(
    df["Time_ms"],
    df["Current_Throughput_Mbps"],
    linewidth=1.0,
    label="Current"
)

ax.plot(
    df["Time_ms"],
    df["Target_Throughput_Mbps"],
    linestyle="--",
    linewidth=1.0,
    label="Target"
)

ax.set_xlabel("Time (ms)")
ax.set_ylabel("Throughput (Mbps)")

ax.grid(
    True,
    linestyle=":",
    linewidth=0.5,
    alpha=0.7
)

ax.legend(
    frameon=False
)

fig.tight_layout()

save_ieee(
    fig,
    "Fig7_Throughput"
)

plt.show()


# ============================================================
# FIGURE 8
# PACKET LOSS
# ============================================================

fig, ax = plt.subplots(
    figsize=(3.45, 2.5)
)

ax.plot(
    df["Time_ms"],
    df["Packet_Loss_pct"],
    linewidth=1.0
)

ax.set_xlabel("Time (ms)")
ax.set_ylabel("Packet Loss (%)")

ax.grid(
    True,
    linestyle=":",
    linewidth=0.5,
    alpha=0.7
)

fig.tight_layout()

save_ieee(
    fig,
    "Fig8_Packet_Loss"
)

plt.show()


# ============================================================
# FIGURE 9
# REWARD
# ============================================================

fig, ax = plt.subplots(
    figsize=(3.45, 2.5)
)

ax.plot(
    df["Time_ms"],
    df["Reward"],
    linewidth=0.8
)

ax.set_xlabel("Time (ms)")
ax.set_ylabel("Reward")

ax.grid(
    True,
    linestyle=":",
    linewidth=0.5,
    alpha=0.7
)

fig.tight_layout()

save_ieee(
    fig,
    "Fig9_DQN_Reward"
)

plt.show()


# ============================================================
# FIGURE 10
# RLF / HANDOVER EVENTS
# ============================================================

fig, ax = plt.subplots(
    figsize=(3.45, 2.2)
)

handover = df["Handover_Flag"]

ax.plot(
    df["Time_ms"],
    handover,
    linewidth=0.9
)

ax.set_xlabel("Time (ms)")
ax.set_ylabel("Handover Flag")

ax.set_yticks([0, 1])

ax.grid(
    True,
    linestyle=":",
    linewidth=0.5,
    alpha=0.7
)

fig.tight_layout()

save_ieee(
    fig,
    "Fig10_Handover_Events"
)

plt.show()


# ============================================================
# FINAL REPORT
# ============================================================

print("\n" + "=" * 65)
print("AKASH-6G IEEE FIGURE GENERATION COMPLETE")
print("=" * 65)

print("\nGenerated figures:")

for file in sorted(os.listdir(OUTPUT_DIR)):
    print("  ", file)

print("\nOutput directory:")
print(OUTPUT_DIR)

print("=" * 65)
```

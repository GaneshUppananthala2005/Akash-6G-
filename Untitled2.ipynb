```python
import numpy as np

def compute_reward(sinr_t, current_action, prev_action, sinr_threshold=5.0):
    """
    Implements Equation (1) from the proposal.
    """
    w1, w2, w3 = 1.0, 0.5, 2.0  # Weights to balance throughput vs. handover cost
    C_HO = 10  # Signaling cost
    P_fail = 50  # Severe penalty for RLF

    # 1. Throughput Gain (Shannon capacity)
    # Ensure 1 + sinr_t is always positive to avoid log(0) or log(negative)
    throughput_gain = np.log2(np.maximum(1e-9, 1 + sinr_t))

    # 2. Switching Penalty (only if action changed)
    switching_penalty = C_HO if current_action != prev_action else 0

    # 3. Resilience Drop Penalty (RLF avoidance)
    resilience_penalty = P_fail if sinr_t < sinr_threshold else 0

    # Total Reward
    reward = (w1 * throughput_gain) - (w2 * switching_penalty) - (w3 * resilience_penalty)
    return reward
```

```python
import matplotlib.pyplot as plt

# Generate a range of sinr_t values
sinr_values = np.linspace(-10, 20, 100) # From -10 dB to 20 dB

# Initialize lists to store rewards
rewards = []

# Assume no action change for this visualization (current_action == prev_action)
current_action = 0
prev_action = 0

# Calculate reward for each sinr_t value
for sinr_t in sinr_values:
    reward = compute_reward(sinr_t, current_action, prev_action)
    rewards.append(reward)

# Plot the results
plt.figure(figsize=(10, 6))
plt.plot(sinr_values, rewards)
plt.xlabel('SINR (sinr_t)')
plt.ylabel('Reward')
plt.title('Reward vs. SINR (No Action Change)')
plt.grid(True)
plt.show()
```
